```python
'''Q3.Write a Python program to accept a string and:Count the number of characters,
Count vowels and consonants,Reverse the string,Check whether it is a palindrome.'''

def is_vowel(ch):
    """Check if a character is a vowel (case-insensitive)."""
    return ch.lower() in 'aeiou'

string = input("Enter a string: ") #user input for the string example: "Never odd or even"

cleaned_str = ''.join(ch for ch in string if ch.isalpha()) #removes spaces

total_chars = len(string)# Count total characters (including spaces and punctuation)

# Count vowels and consonants
vowels_count = sum(1 for ch in cleaned_str if is_vowel(ch))
consonants_count = sum(1 for ch in cleaned_str if ch.isalpha() and not is_vowel(ch))

# Reverse the string
reversed_str = string[::-1]

# Check palindrome (case-insensitive, ignoring spaces and punctuation)
is_palindrome = cleaned_str.lower() == cleaned_str[::-1].lower()

# Display results
print("Number of characters (including spaces):", total_chars)
print("Number of vowels:", vowels_count)
print("Number of consonants:", consonants_count)
print("Reversed string:", reversed_str)
print("Palindrome check:", "Yes" if is_palindrome else "No")
```

```python
'''Q4. Create two sets of integers and demonstrate union, intersection, difference, and 
symmetric difference. Also create a dictionary containing the names and marks of five 
students and display the student with the highest marks.'''
int_set1 = {1, 7, 3, 6, 5}
int_set2 = {9, 5, 4, 7, 8}

# set operations
print("Union:", int_set1 | int_set2)
print("Intersection:", int_set1 & int_set2)
print("Difference (Set1 - Set2):", int_set1 - int_set2)
print("Symmetric Difference:", int_set1 ^ int_set2)

# dictionary of students and their marks
students = {"Alice": 85,"Bob": 90,"Charlie": 78,"Diana": 95,"Eve": 88}

# student with the highest marks
highest_student = max(students, key=students.get)
print("Student with the highest marks:", highest_student, "with marks:", students[highest_student])
```

```python
'''Q5. Develop a Python program that accepts the grade/grade point and 
credit of different courses and calculates the student's SGPA/GPA'''
def calculate_sgpa(courses):
    total_gpoints = 0
    total_credits = 0
    for course in courses:
        grade_point, credit = course
        total_gpoints += grade_point * credit
        total_credits += credit
    sgpa = total_gpoints / total_credits if total_credits > 0 else 0
    return sgpa

courses = [(8, 3), (9, 4), (7, 2), (9, 3), (6,2)]
sgpa = calculate_sgpa(courses)
print("The student's SGPA is:", sgpa)
```

```python
'''Q6.A.Calculate the factorial of a number.
B.Generate the Fibonacci series up to n terms.
C.Check whether a number is prime.'''
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    elif n == 0 or n == 1:
        return 1
    else:
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result

print(factorial(7))
print(factorial(0))
print(factorial(-5))

def fibonacci(n): 
    fib_series = []
    a, b = 0, 1
    for _ in range(n):
        fib_series.append(a)
        a, b = b, a + b
    return fib_series

print(fibonacci(10))  

def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

print(is_prime(29))
print(is_prime(15))

```

```python
'''Q7. Write a Python program to create a text file, write student details into it, and subsequently read and display the contents of the file.'''
def write_student_details(filename, student_details):
    with open(filename, 'w') as file:
        for student in student_details:
            file.write(f"{student['name']},{student['marks']}\n")   
            
def read_student_details(filename):
    with open(filename, 'r') as file:
        for line in file:
            print(line.strip())

student_details = [
    {'name': 'Rahul', 'marks': 85}, 
    {'name': 'Doll', 'marks': 90},
    {'name': 'Anna', 'marks': 78},
    {'name': 'Sam', 'marks': 95},
    {'name': 'Mihir', 'marks': 88}
]

write_student_details('student_details.txt', student_details)
read_student_details('student_details.txt')
```

```python
'''Q8. Write a program that accepts two numbers and performs division using tryexcept. 
Handle the division-by-zero exception appropriately.'''
def divide_numbers(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed."    

a = divide_numbers(10, 2)
b = divide_numbers(10, 0)
print("Result of division (10 / 2):", a)
print("Result of division (10 / 0):", b)
```

```python
'''Q9.Using list comprehension, generate:Squares of numbers from 1 to 20.
Even numbers from 1 to 50. Numbers divisible by both 3 and 5 from 1 to 100.'''
squares = [x**2 for x in range(1, 21)]
even_numbers = [x for x in range(1, 51) if x % 2 == 0]
divisible_by_3_and_5 = [x for x in range(1, 101) if x % 3 == 0 and x % 5 == 0]
print("Squares of numbers from 1 to 20:", squares)
print("Even numbers from 1 to 50:", even_numbers)       
print("Numbers divisible by both 3 and 5 from 1 to 100:", divisible_by_3_and_5)
```

```python
# Student record management using dictionary

# Empty dictionary to store student names and marks
students = {}

# Display menu options
print("\n1. Add Student\n2. Update Marks\n3. Search Student\n4. Delete Student\n5. Display All Students\n6. Exit")

# Loop to keep displaying the menu until the user exits
while True:

    # Take the user's choice
    choice = int(input("Enter your choice: "))

    # Add a new student and their marks to the dictionary
    if choice == 1:
        name = input("Enter student name: ")
        marks = int(input("Enter marks: "))
        students[name] = marks
        print("Student added successfully!",name, "with marks:", marks)

    # Update the marks of an existing student
    elif choice == 2:
        name = input("Enter student name to update: ")

        # Check if the student exists in the dictionary
        if name in students:
            marks = int(input("Enter new marks: "))
            students[name] = marks
            print("Marks updated successfully!")
        else:
            print("Student not found!")

    # Search for a student and display their marks
    elif choice == 3:
        name = input("Enter student name to search: ")

        # Display the record if the student exists
        if name in students:
            print("Student Name:", name)
            print("Marks:", students[name])
        else:
            print("Student not found!",name)

    # Delete a student record from the dictionary
    elif choice == 4:
        name = input("Enter student name to delete: ")

        # Check if the student exists before deleting
        if name in students:
            del students[name]
            print("Student deleted successfully!")
        else:
            print("Student not found!",name)

    # Display all stored student records
    elif choice == 5:
        print("Student Records:", students)

    # Exit the program
    elif choice == 6:
        print("Exiting program...")
        break

    # Handle an invalid menu choice
    else:
        print("Invalid choice!")
```
