```python
# Recursive factorial
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print("Factorial of 5:", factorial(5))

# Recursive sum of first n natural numbers
def natural_sum(n):
    if n == 0:
        return 0
    return n + natural_sum(n - 1)

print("Sum of first 5 natural numbers:", natural_sum(5))

# Recursive power for a non-negative exponent
def power(x, n):
    if n == 0:
        return 1
    return x * power(x, n - 1)

print("2 raised to 5:", power(2, 5))
```

```python
# Reverse a sequence in place using recursion
def reverse_sequence(data, start, stop):
    if start >= stop:
        return
    data[start], data[stop] = data[stop], data[start]
    reverse_sequence(data, start + 1, stop - 1)

sequence = [1, 2, 3, 4, 5, 6]
reverse_sequence(sequence, 0, len(sequence) - 1)
print("Reversed sequence:", sequence)
```

```python
# Recursive check for non-decreasing order
def is_sorted(data, index=0):
    if index >= len(data) - 1:
        return True
    if data[index] > data[index + 1]:
        return False
    return is_sorted(data, index + 1)

print(is_sorted([1, 2, 2, 4, 5]))
print(is_sorted([1, 3, 2, 4]))

```

```python
# Naive recursive Fibonacci
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print("Fibonacci(5):", fibonacci(5))
print("First 6 Fibonacci values:", [fibonacci(i) for i in range(6)])

```

```python
# Tower of Hanoi recursive solution
def tower_of_hanoi(n, source, auxiliary, destination):
    if n == 1:
        print("Move disk 1 from", source, "to", destination)
        return
    tower_of_hanoi(n - 1, source, destination, auxiliary)
    print("Move disk", n, "from", source, "to", destination)
    tower_of_hanoi(n - 1, auxiliary, source, destination)

tower_of_hanoi(3, "A", "B", "C")

# Tail-recursive sum and its iterative conversion
def linear_sum_tail(n, total=0):
    if n == 0:
        return total
    return linear_sum_tail(n - 1, total + n)

def linear_sum_iterative(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

print("Tail-recursive result:", linear_sum_tail(5))
print("Iterative result:", linear_sum_iterative(5))
```

```python
# Stack implementation using a Python list
class Stack:
    def __init__(self):
        self.data = []

    def push(self, item):
        self.data.append(item)

    def pop(self):
        if len(self.data) == 0:
            raise IndexError("Stack is empty")
        return self.data.pop()

    def top(self):
        if len(self.data) == 0:
            raise IndexError("Stack is empty")
        return self.data[-1]

    def is_empty(self):
        return len(self.data) == 0

    def __len__(self):
        return len(self.data)

stack = Stack()
stack.push(10)
stack.push(20)
stack.push(30)
print("Top:", stack.top())
print("Popped:", stack.pop())
print("Stack length:", len(stack))

# Reverse a sequence using a stack
def reverse_using_stack(sequence):
    stack = []
    for item in sequence:
        stack.append(item)

    result = []
    while stack:
        result.append(stack.pop())
    return result

print("Reversed sequence:", reverse_using_stack([1, 2, 3, 4, 5]))
```

```python
# Circular-array queue implementation
class CircularQueue:
    def __init__(self, capacity):
        self.data = [None] * capacity
        self.capacity = capacity
        self.size = 0
        self.front = 0

    def enqueue(self, item):
        if self.size == self.capacity:
            raise IndexError("Queue is full")
        index = (self.front + self.size) % self.capacity
        self.data[index] = item
        self.size += 1

    def dequeue(self):
        if self.size == 0:
            raise IndexError("Queue is empty")
        item = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return item

    def first(self):
        if self.size == 0:
            raise IndexError("Queue is empty")
        return self.data[self.front]

    def is_empty(self):
        return self.size == 0

    def __len__(self):
        return self.size

queue = CircularQueue(5)
queue.enqueue("A")
queue.enqueue("B")
queue.enqueue("C")
print("Front item:", queue.first())
print("Dequeued:", queue.dequeue())
print("Queue size:", len(queue))
```

```python
# Deque operations using circular-array indexing
class ArrayDeque:
    def __init__(self, capacity):
        self.data = [None] * capacity
        self.capacity = capacity
        self.size = 0
        self.front = 0

    def add_first(self, item):
        if self.size == self.capacity:
            raise IndexError("Deque is full")
        self.front = (self.front - 1) % self.capacity
        self.data[self.front] = item
        self.size += 1

    def add_last(self, item):
        if self.size == self.capacity:
            raise IndexError("Deque is full")
        index = (self.front + self.size) % self.capacity
        self.data[index] = item
        self.size += 1

    def delete_first(self):
        if self.size == 0:
            raise IndexError("Deque is empty")
        item = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return item

    def delete_last(self):
        if self.size == 0:
            raise IndexError("Deque is empty")
        index = (self.front + self.size - 1) % self.capacity
        item = self.data[index]
        self.data[index] = None
        self.size -= 1
        return item

    def display(self):
        return [self.data[(self.front + i) % self.capacity] for i in range(self.size)]

deque = ArrayDeque(5)
deque.add_first(20)
deque.add_first(10)
deque.add_last(30)
print("Deque:", deque.display())
print("Deleted first:", deque.delete_first())
print("Deleted last:", deque.delete_last())
print("Remaining:", deque.display())
```

```python
# Singly linked list node and required operations
class Node:
    def __init__(self, element):
        self.element = element
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def insert_first(self, element):
        node = Node(element)
        node.next = self.head
        self.head = node
        if self.tail is None:
            self.tail = node
        self.size += 1

    def insert_last(self, element):
        node = Node(element)
        if self.head is None:
            self.head = self.tail = node
        else:
            self.tail.next = node
            self.tail = node
        self.size += 1

    def delete_first(self):
        if self.head is None:
            raise IndexError("List is empty")
        element = self.head.element
        self.head = self.head.next
        self.size -= 1
        if self.head is None:
            self.tail = None
        return element

    def traversal(self):
        elements = []
        current = self.head
        while current is not None:
            elements.append(current.element)
            current = current.next
        return elements

linked = SinglyLinkedList()
linked.insert_first(20)
linked.insert_first(10)
linked.insert_last(30)
print("List:", linked.traversal())
print("Deleted:", linked.delete_first())
print("After deletion:", linked.traversal())
```

```python
# Favorites List ADT supporting access, remove and top(k)
class FavoritesList:
    def __init__(self):
        self.items = []

    def access(self, element):
        for item in self.items:
            if item["element"] == element:
                item["count"] += 1
                return
        self.items.append({"element": element, "count": 1})

    def remove(self, element):
        self.items = [item for item in self.items if item["element"] != element]

    def top(self, k):
        ranked = sorted(self.items, key=lambda item: item["count"], reverse=True)
        return ranked[:k]

favorites = FavoritesList()
favorites.access("Python")
favorites.access("Python")
favorites.access("Algorithms")
favorites.access("Python")
favorites.access("Data Structures")
favorites.access("Algorithms")
print("Top 2:", favorites.top(2))
favorites.remove("Data Structures")
print("After removal:", favorites.items)
```

```python
# Doubly linked list with insertion and deletion operations
class DoublyNode:
    def __init__(self, element):
        self.element = element
        self.prev = None
        self.next = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def insert_first(self, element):
        node = DoublyNode(element)
        node.next = self.head
        if self.head is not None:
            self.head.prev = node
        else:
            self.tail = node
        self.head = node
        self.size += 1

    def insert_last(self, element):
        node = DoublyNode(element)
        node.prev = self.tail
        if self.tail is not None:
            self.tail.next = node
        else:
            self.head = node
        self.tail = node
        self.size += 1

    def delete_first(self):
        if self.head is None:
            raise IndexError("List is empty")
        element = self.head.element
        self.head = self.head.next
        if self.head is not None:
            self.head.prev = None
        else:
            self.tail = None
        self.size -= 1
        return element


    def delete_last(self):
        if self.tail is None:
            raise IndexError("List is empty")
        element = self.tail.element
        self.tail = self.tail.prev
        if self.tail is not None:
            self.tail.next = None
        else:
            self.head = None
        self.size -= 1
        return element


    def display_forward(self):
        elements = []
        current = self.head
        while current is not None:
            elements.append(current.element)
            current = current.next
        return elements


    def display_backward(self):
        elements = []
        current = self.tail
        while current is not None:
            elements.append(current.element)
            current = current.prev
        return elements


# Demonstrate insertion and deletion at both ends
items = DoublyLinkedList()
items.insert_first(20)
items.insert_first(10)
items.insert_last(30)
print("Forward traversal:", items.display_forward())
print("Backward traversal:", items.display_backward())
print("Deleted first:", items.delete_first())
print("Deleted last:", items.delete_last())
print("After deletions:", items.display_forward())
```
